## 00 - Etatística
### Conceitos básicos de estatística
- População: conjunto completo de elementos que possuem caracteristica estudada.
- Amostra: subconjunto utilizado para inferir sobre a população.
- Parâmetro: uma característica numérica da pupulação (média, variância, etc..).
- Estatística: uma quantidade calculada a partir da amostra. (média amostral, variância amostral, etc...).

Modelos de ML quase sempre treinam com amostra generalizando para população.

---
### Distribuições e Suficiência Estatística

- Ruído: parte aleatórioa, imprevisível ou irrelevante dos dados que atrapalha enxergar o padrão real.
- Sinal: estrutura real, padrão verdadeiro.
- Dados = sinal + ruído
- Regressão linear: Modela relação entre variáveis através de uma reta (prever uma variável utilizando outra). Erros aproximadamente normais facilitam a inferência de coeficientes.
- Classificação probabilística: modela-se distribuição condicional das classes. Ex: gato(92%), cachorro(7%), raposa(1%).
Distribuição empírica é uma aproximação computacional da distribuição real. O histograma é uma discretização com perda de informação (trade-off resolução vs ruído).
Quanto maior a amostra, mais a distribuição empírica se aproxima da distribuição real.

---
### Demonstração incremental: Lei dos Grandes Números (LLN)
A LLN explica por que a média amostral converge para a média populacional com o crescimento.

### Teorema Central do limite (CLT)
Mesmo com dados individuais não normais, a distribuição média amostral tende a normalidade com n suficientemente grande.

### Inferência: Intervalo de Confiança da Média
É um intervalo construído a partar da amostra para estimar a média verdadeira da população.

### Teste de Hipótese e Erros Tipo I/II
- H0: hipótese nula
- H1: hipótese alternativa
- p-valor: probabilidade de observar estatística tão extrema quanto a observada, assumindo H0 verdadeira.

### Correlação vs Causalidade
Correlação mede associação linear. Alta correlação pode ajudar a prever causalidade, mas não necessariamente.

### Viés, Variância e Erro de generalização
- Viés: SImplificação excessiva do modelo
- Variância: Sensibilidade ao ruído da amostra
- Ruído irreducível 

### Métricas Base: Regressão e Classificação
Escolha de métrica é decisão de negócio:

    Regressão: MAE, RMSE, R²
    Classificação: precision, recall, F1, ROC-AUC, PR-AUC

### Erros comuns e como deupurar

- Confundir média de amostra com verdade populacional
- Usar teste paramétrico sem verificar premissas
- Escolher métrica inadequada para desbalanceamento
- Interpretar correlação como causalidade

Checklist de depuração estatística

- Verificar qualidade dos dados e outliers
- Conferir distribuição e escala
- Validar suposições do método
- Reportar incerteza (intervalos, variabilidade)
